```python
import pickle, json
import numpy as np

with open('Random_Forest_pipeline.pkl', 'rb') as f:
    pipe = pickle.load(f)

scaler = pipe.named_steps['scaler']

with open('model_metadata.json') as f:
    meta = json.load(f)

export = {
    'feature_names': meta['feature_names'],  # exact ordered list
    'scaler_mean':   scaler.mean_.tolist(),
    'scaler_std':    scaler.scale_.tolist(),  # std, not var
    'n_features':    int(scaler.n_features_in_),
    'img_size':      [256, 256],
    'grid_size':     3,
    'cell_size':     [85, 85],
    'canny_low':     50,
    'canny_high':    150,
    'lbp_P':         8,
    'lbp_R':         1.0,
    'lbp_bins':      10,
    'low_edge_thresh': 0.05,
}

with open('swift_inference_params.json', 'w') as f:
    json.dump(export, f, indent=2)

print("Exported swift_inference_params.json")
```
